### **불필요한 컬럼 삭제**

```python
drop_cols = ["trans_num", "cc_num", "first", "last", "trans_dt", "street", "city", "zip", "job", "merchant", "unix_time"]
cc_df = cc_df.drop(columns=drop_cols, errors="ignore")
cc_df.shape
```

### **컬럼 이름 통일**

```python
cc_df = cc_df.rename(columns={
    "lat": "cust_lat",
    "long": "cust_lon",
    "merch_lat": "merch_lat",
    "merch_long": "merch_lon",
    "trans_date_trans_time": "trans_dt"
})
cc_df.head()
```

### **고객 생일을 나이로 계산**

```python
cc_df.head()
```

```python
cc_df["trans_dt"] = pd.to_datetime(cc_df["trans_dt"], errors="coerce")
```

```python
cc_df['dob_dt'] = pd.to_datetime(cc_df['dob'], errors= 'coerce')
```

```python
cc_df["age"] = (cc_df["trans_dt"] - cc_df["dob_dt"]).dt.days // 365
```

### **거래 시간대**

```python
hour, weekday/weekend
cc_df['trans_dt'] = pd.to_datetime(cc_df['trans_dt'])
cc_df['hour'] = cc_df['trans_dt'].dt.hour
cc_df['weekday'] = cc_df['trans_dt'].dt.weekday   # 0=월, 6=일
cc_df['is_weekend'] = cc_df['weekday'].isin([5,6]).astype(int)
```
